## 🧠 Parte 2: El Arte del Prompting

⏱️ *Tiempo estimado: 30 minutos*

### ¿Qué es un Prompt?

Un **prompt** es la instrucción o pregunta que le das al modelo de IA. La calidad de tu prompt determina la calidad de la respuesta.

> 💡 **Analogía:** Piensa en el chatbot como un asistente muy inteligente pero que no conoce tu contexto. Cuanto más claro le expliques qué necesitas, mejor te ayudará.

### 🔄 Metodología de Trabajo: El Ciclo de Análisis con IA

```
1️⃣ PLANIFICAR → ¿Qué quiero analizar?
      ⬇️
2️⃣ EJEMPLIFICAR → Mostrar datos de ejemplo al LLM
      ⬇️
3️⃣ GENERAR → Pedir código con prompts claros
      ⬇️
4️⃣ EJECUTAR → Correr el código en Colab
      ⬇️
5️⃣ INTERPRETAR → ¿Tiene sentido el resultado?
      ⬇️
6️⃣ ITERAR → Refinar y repetir
```

### 📝 Anatomía de un Buen Prompt

Un prompt efectivo tiene estos componentes:

| Componente | Descripción | Ejemplo |
|------------|-------------|--------|
| **Contexto** | Describe la situación | "Tengo un dataset de películas de Netflix..." |
| **Datos de ejemplo** | Muestra la estructura | "Las columnas son: title, type, release_year..." |
| **Tarea específica** | Qué quieres lograr | "Quiero saber cuántas películas hay por año" |
| **Formato deseado** | Cómo quieres el resultado | "Dame el código en Python usando pandas" |

### 🎯 Ejemplo: Prompt Malo vs Prompt Bueno

#### ❌ Prompt MALO:
```
"Analiza mis datos"
```

**Problemas:** No hay contexto, no hay datos de ejemplo, no hay objetivo claro.

---

#### ✅ Prompt BUENO:
```
Tengo un dataset de Netflix con las siguientes columnas:
- show_id: identificador único
- type: 'Movie' o 'TV Show'
- title: nombre del contenido
- release_year: año de lanzamiento
- duration: duración (ej: '90 min' o '2 Seasons')

Ejemplo de las primeras filas:
| show_id | type  | title      | release_year | duration |
|---------|-------|------------|--------------|----------|
| s1      | Movie | The Matrix | 1999         | 136 min  |
| s2      | TV Show | Friends  | 1994         | 10 Seasons |

TAREA: Genera código en Python con pandas para:
1. Contar cuántas películas vs series hay
2. Mostrar el resultado como un gráfico de barras
```

## 🎯 Parte 2.5: Business Question Canvas

⏱️ *Tiempo en vivo: 10 minutos (versión completa como práctica opcional)*

### ¿Qué es el Business Question Canvas?

Antes de pedirle a la IA que analice datos, debes tener **claridad sobre qué quieres saber**. El **Business Question Canvas** es una plantilla simple para transformar preguntas vagas en preguntas de negocio estructuradas.

> 💡 **Analogía:** Antes de construir una casa, necesitas planos. Antes de analizar datos, necesitas un Business Question Canvas.

### 📋 Los 6 Elementos de una Pregunta de Negocio

| Elemento | Descripción | Ejemplo |
|----------|-------------|---------|
| **1. Problema** | ¿Qué problema quieres resolver? | "Las ventas cayeron 15% el último trimestre" |
| **2. Pregunta** | ¿Qué quieres saber exactamente? | "¿Qué categoría de productos tiene menor rentabilidad?" |
| **3. Métrica** | ¿Qué número medirá el éxito? | Margen % = (Precio - Costo) / Precio × 100 |
| **4. Audiencia** | ¿Quién tomará decisiones? | Dueño de PyME, Director de Producto |
| **5. Decisión** | ¿Qué acción se tomará? | Ajustar precios, eliminar productos |
| **6. Riesgo** | ¿Qué pasa si te equivocas? | Pérdida de revenue, insatisfacción de clientes |

### ✅ Ejemplo: Pregunta Mal formulada vs Bien formulada

#### ❌ Pregunta MAL formulada:
```
"Analiza los datos de ventas"
```
**Problemas:** No hay contexto, no hay objetivo claro, no hay decisión asociada.

---

#### ✅ Pregunta BIEN formulada (con Canvas):

**1. Problema:** Los márgenes de ganancia han caído del 18% al 12% en el último año. No sabemos qué productos son más rentables.

**2. Pregunta:** ¿Cuáles son los 5 productos con menor margen de ganancia y deberíamos considerar aumentar precio o discontinuar?

**3. Métrica:** Margen % = (Precio - Costo) / Precio × 100

**4. Audiencia:** Dueño de la PyME y Director Comercial

**5. Decisión:** Discontinuar productos con margen < 10% y bajo volumen; Aumentar precio 10-15% en productos con margen 10-13%

**6. Riesgo:** Si eliminamos los productos equivocados, podemos perder clientes y reducir revenue total.

## 🔍 Parte 2.6: AI Reliability Checklist

⏱️ *Tiempo en vivo: 5 minutos (aplica un check; la lista completa queda como referencia)*

### La Regla de Oro del Análisis con IA

> **"Confía, pero verifica"**
>
> La IA puede alucinar: inventar datos, interpretar mal código, o hacer cálculos incorrectos.
> **Tú eres el piloto**, la IA es tu copiloto. Siempre verifica lo que genera.

### ✅ The 3-Check Method

Para cualquier análisis importante, haz **3 verificaciones**:

#### Check 1: Summary Statistics
```python
# Antes de aceptar un resultado, pide ver:
df.describe()  # Estadísticas resumen
df.info()      # Estructura de datos
```
**Qué verificar:**
- [ ] Los totales tienen sentido (no demasiado altos/bajos)
- [ ] No hay valores imposibles (edad: 200, satisfacción: 7/5)
- [ ] Los tipos de datos son correctos

#### Check 2: Sample Rows
```python
# Revisa las primeras filas del resultado:
df.head(10)  # Primeras 10 filas
```
**Qué verificar:**
- [ ] Los valores se ven correctos visualmente
- [ ] No hay NaN o valores extraños (inf, null)
- [ ] Los cálculos por fila tienen sentido

#### Check 3: Triangulation
```python
# Calcula el mismo resultado de otra forma:
# Método 1
result1 = df.groupby('categoria')['ventas'].sum()

# Método 2 (diferente)
result2 = df.pivot_table(index='categoria', values='ventas', aggfunc='sum')

# Deben dar el mismo resultado
assert result1.equals(result2)  # Si no, investiga por qué
```
**Qué verificar:**
- [ ] Diferentes métodos dan el mismo resultado
- [ ] Si hay diferencia, la IA la explica lógicamente
- [ ] No hay contradicciones en las respuestas

---

### 🚨 Signos de Alucinación

Señales de que la IA podría estar inventando resultados:

🚨 **Números demasiado redondos**
- "El promedio es exactamente 50.0%"
- "Tienen 1,000 clientes exactamente"

🚨 **Resultados demasiado perfectos**
- "Todos los productos tienen el mismo margen"

🚨 **Falta de transparencia**
- No muestra el código que usó
- No puede explicar cómo llegó al resultado

🚨 **Contradicciones**
- Dice "El promedio es 4.2/5" pero el histograma muestra mayoría de 1-2

````python
# SOLUCIONES - Ejecuta después de intentar tú mismo

print("=" * 50)
print("🔍 SOLUCIONES: Detective de Datos")
print("=" * 50)

# Solución 1: Valores faltantes
print("\n1️⃣ VALORES FALTANTES por columna:")
print("-" * 30)
faltantes = df.isnull().sum()
porcentaje = (faltantes / len(df) * 100).round(1)
for col in faltantes[faltantes > 0].index:
    print(f"  {col}: {faltantes[col]} ({porcentaje[col]}%)")

# Solución 2: Filas duplicadas
print("\n2️⃣ FILAS DUPLICADAS:")
print("-" * 30)
duplicados = df.duplicated().sum()
print(f"  Total de filas duplicadas: {duplicados}")
if duplicados > 0:
    print(f"  ⚠️  Esto representa el {duplicados/len(df)*100:.1f}% del dataset")
else:
    print(f"  ✅ No hay filas duplicadas")

# Solución 3: Outliers
print("\n3️⃣ OUTLIERS en release_year:")
print("-" * 30)
anio_min = df['release_year'].min()
anio_max = df['release_year'].max()
print(f"  Año más antiguo: {anio_min}")
print(f"  Año más reciente: {anio_max}")
if anio_min < 1925 or anio_max > 2025:
    print(f"  ⚠️  POSIBLE OUTLIER: Hay años fuera del rango esperado (1925-2025)")

# Solución 4: Consistencia de categorías
print("\n4️⃣ CONSISTENCIA de categorías:")
print("-" * 30)
# Mostrar algunos valores de country
paises_muestra = df['country'].dropna().head(20).tolist()
print(f"  Muestra de valores en 'country':")
for p in paises_muestra:
    print(f"    - '{p}'")
````

## 🧹 Parte 5: Limpieza Básica de Datos

### 5.1 Identificar valores faltantes

Los valores faltantes (NaN) son muy comunes en datasets reales. Es importante identificarlos **antes** de hacer cualquier análisis serio.

> 💡 **¿Por qué importan los valores faltantes?** Pueden distorsionar promedios, romper gráficos, o hacer que tus conclusiones sean incorrectas.

````python
# Contar valores faltantes por columna
valores_faltantes = df.isnull().sum()

# Mostrar solo las columnas con valores faltantes
print("📊 Columnas con valores faltantes:")
print("=" * 40)
print(valores_faltantes[valores_faltantes > 0].sort_values(ascending=False))
````

### 5.2 Visualizar valores faltantes

````python
# Crear gráfico de valores faltantes
plt.figure(figsize=(10, 5))

# Calcular porcentaje de valores faltantes
porcentaje_faltantes = (df.isnull().sum() / len(df) * 100).sort_values(ascending=True)

# Crear gráfico de barras horizontales
porcentaje_faltantes.plot(kind='barh', color='coral')

plt.title('📊 Porcentaje de Valores Faltantes por Columna', fontsize=14, fontweight='bold')
plt.xlabel('Porcentaje (%)')
plt.ylabel('Columna')

# Agregar línea vertical en 0%
plt.axvline(x=0, color='black', linewidth=0.5)

plt.tight_layout()
plt.show()
````

## 📈 Parte 6: Análisis Básico con Prompting

### 6.1 Análisis por Año de Lanzamiento

Vamos a analizar cómo se distribuyen los títulos por año de lanzamiento.

````python
# Contar títulos por año de lanzamiento
titulos_por_año = df['release_year'].value_counts().sort_index()

# Filtrar últimos 20 años para mejor visualización
titulos_recientes = titulos_por_año[titulos_por_año.index >= 2005]

# Crear gráfico
plt.figure(figsize=(12, 5))
titulos_recientes.plot(kind='bar', color='steelblue', edgecolor='white')

plt.title('📊 Títulos de Netflix por Año de Lanzamiento (2005-2021)', fontsize=14, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Cantidad de Títulos')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
````

### 6.2 Comparación: Películas vs Series por Año

````python
# Agrupar por año y tipo
por_año_tipo = df[df['release_year'] >= 2005].groupby(['release_year', 'type']).size().unstack(fill_value=0)

# Crear gráfico
fig, ax = plt.subplots(figsize=(12, 5))

por_año_tipo.plot(kind='bar', ax=ax, width=0.8)

plt.title('📊 Películas vs Series por Año de Lanzamiento', fontsize=14, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Cantidad de Títulos')
plt.legend(title='Tipo')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
````

## 🎯 Parte 7: Técnicas Avanzadas de Prompting

### 7.1 El Prompt de "Contexto Completo"

Cuando necesites hacer análisis más complejos, usa este formato:

```
CONTEXTO:
[Describe la situación y los datos]

DATOS:
[Muestra ejemplos específicos de tus datos]

OBJETIVO:
[Qué quieres lograr]

RESTRICCIONES:
[Cualquier limitación o preferencia]

FORMATO DE SALIDA:
[Cómo quieres el código o resultado]
```

### 7.2 Ejemplo de Prompt Completo:

```
CONTEXTO:
Soy analista de datos en una empresa de streaming competidora de Netflix.
Necesito entender qué géneros dominan el catálogo de Netflix para nuestra estrategia.

DATOS:
Tengo un DataFrame 'df' con una columna 'listed_in' que contiene los géneros.
Ejemplos de valores:
- 'Documentaries'
- 'International Movies, Dramas, Comedies'
- 'TV Dramas, International TV Shows'

Nota: Una película puede tener múltiples géneros separados por coma.

OBJETIVO:
1. Separar los géneros (split por coma)
2. Contar la frecuencia de cada género individual
3. Mostrar los 15 géneros más populares

RESTRICCIONES:
- Usar pandas y matplotlib
- El gráfico debe tener título y etiquetas claras

FORMATO DE SALIDA:
Código Python comentado que pueda ejecutar en Google Colab
```

### 7.3 Ejecutemos este análisis de géneros:

````python
# Análisis de géneros en Netflix

# Paso 1: Separar los géneros (cada título puede tener varios)
# Usamos explode() para crear una fila por cada género
generos = df['listed_in'].str.split(', ').explode()

# Paso 2: Contar la frecuencia de cada género
conteo_generos = generos.value_counts().head(15)

# Paso 3: Crear el gráfico
plt.figure(figsize=(12, 6))
conteo_generos.plot(kind='barh', color='teal', edgecolor='white')

plt.title('🎬 Top 15 Géneros más Populares en Netflix', fontsize=14, fontweight='bold')
plt.xlabel('Cantidad de Títulos')
plt.ylabel('Género')

# Invertir el orden para que el más popular quede arriba
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

# Mostrar los números exactos
print("\n📊 Conteo exacto por género:")
print(conteo_generos)
````
