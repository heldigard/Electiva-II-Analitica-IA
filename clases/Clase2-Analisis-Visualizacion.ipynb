## 🎯 Decision Flow: ¿Qué Gráfico Usar y Cuándo?

⏱️ *Tiempo estimado: 10 minutos*

### El Decision Flow para Gráficos

No todos los gráficos sirven para todas las preguntas. Usa este **decision flow** para elegir el gráfico correcto:

```
¿QUÉ TIPO DE PREGUNTA TIENES?
│
├─ "¿Cómo se distribuyen los valores?"
│   → HISTOGRAMA
│   Ejemplo: ¿Cómo se distribuyen los precios de los productos?
│
├─ "¿Qué categoría es mayor/menor?"
│   → GRÁFICO DE BARRAS
│   Ejemplo: ¿Qué categoría de productos vende más?
│
├─ "¿Cómo cambió algo en el tiempo?"
│   → GRÁFICO DE LÍNEAS
│   Ejemplo: ¿Cómo han evolucionado las ventas mensuales?
│
├─ "¿Hay relación entre X e Y?"
│   → SCATTER PLOT
│   Ejemplo: ¿Más precio = mejor calificación?
│
└─ "¿Hay valores atípicos?"
    → BOXPLOT
    Ejemplo: ¿Hay productos con precios extremos?
```

### 📊 Decision Flow Detallado

| Tu Pregunta | Variable X | Variable Y | Gráfico Recomendado |
|-------------|------------|------------|---------------------|
| ¿Cuál es la distribución de X? | Numérica | - | Histograma |
| ¿Qué categoría de X tiene mayor Y? | Categórica | Numérica | Barras |
| ¿Cómo cambió X en el tiempo? | Tiempo | Numérica | Líneas |
| ¿X e Y están relacionados? | Numérica | Numérica | Scatter |
| ¿Cuáles son los outliers de X por grupo? | Categórica | Numérica | Boxplot |

### ¿Qué es?
Un histograma muestra **cómo se distribuyen los valores** de una variable numérica. Divide los datos en "barras" (bins) y muestra cuántos valores caen en cada rango.

### ¿Cuándo usarlo?
- Para entender la distribución de precios, edades, cantidades, etc.
- Para identificar si los datos están concentrados o dispersos
- Para detectar si hay múltiples "picos" (modas) en los datos

> 💡 **Pregunta clave que responde:** "¿Cómo se distribuyen los valores de esta variable?"

### Ejemplo: Distribución de Años de Lanzamiento en Netflix

````python
# Histograma de años de lanzamiento
plt.figure(figsize=(12, 5))

# Filtrar años recientes para mejor visualización
años_recientes = df_netflix[df_netflix['release_year'] >= 1990]['release_year']

plt.hist(años_recientes, bins=30, color='steelblue', edgecolor='white', alpha=0.7)

plt.title('📊 Distribución de Títulos por Año de Lanzamiento', fontsize=14, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Cantidad de Títulos')

# Agregar línea vertical en la mediana
mediana = años_recientes.median()
plt.axvline(x=mediana, color='red', linestyle='--', label=f'Mediana: {mediana:.0f}')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\n📈 Estadísticas:")
print(f"   Año promedio: {años_recientes.mean():.1f}")
print(f"   Año mediano: {mediana:.0f}")
print(f"   Año mínimo: {años_recientes.min()}")
print(f"   Año máximo: {años_recientes.max()}")
````

### 🎓 Interpretación del Histograma

**Observaciones que puedes verificar en la salida de la celda anterior:**
- La mediana indica el año que deja la mitad de los títulos a cada lado.
- Compara la media con la mediana y observa dónde se concentran las barras antes de describir el sesgo.
- Este gráfico usa `release_year`: muestra el año de lanzamiento, **no** la cantidad de títulos que Netflix agregó cada año.

**Insight de negocio (redáctalo solo con evidencia):**
> "La mediana de lanzamiento es **[año calculado]** y **[patrón visible]**. Esto describe la composición del catálogo; para explicar una causa o recomendar una inversión necesitaríamos datos adicionales sobre licencias, audiencia o costos."

## 🎯 Parte Extra: EJERCICIO - Insight vs. Observación

⏱️ *Tiempo estimado: 15 minutos*

### ¿Por qué este ejercicio?

Mucha gente describe lo que ve en un gráfico, pero pocos extraen **insights de negocio**.

> **La diferencia clave:**
> - **Observación:** Describe lo que hay (hechos)
> - **Insight:** Interpreta qué significa para el negocio (impacto, acción)

### Ejemplo Práctico

Mira este ejemplo con un gráfico de ventas:

❌ **Observación** (poco útil):
> "Las ventas subieron en Q4"

✅ **Insight** (valioso para el negocio):
> "En este caso sintético, las ventas de noviembre y diciembre superan sus metas mensuales.
> Una hipótesis es un efecto estacional; habría que compararlo con otros años antes de atribuirlo a una campaña.
> **Siguiente paso:** revisar ventas históricas y datos de campañas antes de cambiar la inversión."

````python
# Crear un gráfico para practicar Insight vs. Observación

# Datos de ejemplo: Ventas mensuales de 2025
ventas_mensuales = pd.DataFrame({
    'mes': ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 
            'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'],
    'ventas': [120, 135, 128, 142, 138, 150, 145, 155, 165, 180, 280, 320],
    'meta': [140, 140, 140, 145, 145, 145, 150, 150, 150, 160, 200, 200]
})

# Crear gráfico
plt.figure(figsize=(12, 6))
plt.plot(ventas_mensuales['mes'], ventas_mensuales['ventas'], 
         marker='o', linewidth=2, markersize=8, label='Ventas Reales', color='steelblue')
plt.plot(ventas_mensuales['mes'], ventas_mensuales['meta'], 
         marker='s', linewidth=2, markersize=6, linestyle='--', 
         label='Meta Mensual', color='coral')

plt.title('📊 Ventas Mensuales 2025 vs. Meta', fontsize=14, fontweight='bold')
plt.xlabel('Mes')
plt.ylabel('Ventas (miles USD)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(0, 350)
plt.tight_layout()
plt.show()

# Mostrar los datos
print("Datos de ventas:")
print(ventas_mensuales)
````

### 🎓 Desglosando un Buen Insight

Un insight de negocio efectivo tiene:

1. **Evidencia numérica** específica (no "subió", sino "subió X%")
2. **Contexto** de comparación (vs. promedio, vs. meta, vs. año anterior)
3. **Hipótesis explícita** de una causa, diferenciada del dato (por ejemplo: "podría relacionarse con…").
4. **Recomendación accionable** (qué hacer basado en este insight)

**Fórmula mágica:**
```
[Número] + [Comparación] + [Hipótesis por verificar] + [Acción prudente]
```

**Ejemplo:**
```
Ventas de noviembre y diciembre (280 + 320 = 600k)
+ superan la meta de esos meses (200 + 200 = 400k) en 50%
+ Hipótesis: podría existir un efecto estacional; el gráfico por sí solo no identifica la causa
+ Acción: revisar datos de años anteriores antes de cambiar la inversión de Q4
```

## 📖 Parte Extra: Gráfico Narrativo - Cuéntalo con una Historia

⏱️ *Tiempo estimado: 20 minutos*

### ¿Qué es un Gráfico Narrativo?

Un gráfico narrativo es más que un gráfico bonito. **Cuenta una historia** con:

1. **Headline** - Título que comunica el insight principal
2. **Anotaciones** - Flechas y texto explicativos sobre puntos clave
3. **Key Takeaway** - 1 frase con el mensaje principal
4. **Call to Action** - Qué hacer basado en este insight

### 🎯 Por qué importa

Los gráficos tradicionales muestran datos. Los gráficos narrativos **persuaden**.

| Gráfico Tradicional | Gráfico Narrativo |
|---------------------|-------------------|
| "Ventas por mes" | "🎯 Nov–Dic superan su meta en 50% (600k vs 400k)" |
| Muestra datos | Cuenta una historia |
| El lector interpreta | Tú guías la interpretación |
| Pasivo | Activo y persuasivo |

### 📊 Ejemplo: Gráfico Narrativo de Categorías Netflix


````python
# GRÁFICO NARRATIVO: Géneros en Netflix

# Preparar datos
generos = df_netflix['listed_in'].str.split(', ').explode()
top_generos = generos.value_counts().head(10)

# Crear figura
fig, ax = plt.subplots(figsize=(12, 7))

# Crear gráfico de barras horizontales
bars = ax.barh(range(len(top_generos)), top_generos.values, color='teal')

# ============ ELEMENTOS NARRATIVOS ============

# 1. HEADLINE - Título calculado a partir de los datos
genero_top = top_generos.index[0]
porcentaje_top = top_generos.iloc[0] / len(df_netflix) * 100
diferencia_vs_segundo = (top_generos.iloc[0] / top_generos.iloc[1] - 1) * 100
ax.set_title(f'🎯 {genero_top} es el género más frecuente ({porcentaje_top:.1f}% de títulos)',
             fontsize=16, fontweight='bold', pad=20)

# 2. ANOTACIONES - Flechas y texto explicativos
# Resaltar la barra #1
bars[0].set_color('#e74c3c')  # Rojo para destacar
ax.annotate(f'TOP 1\n{diferencia_vs_segundo:.0f}% más que #2',
            xy=(top_generos.values[0], 0),
            xytext=(top_generos.values[0] + 200, 0.5),
            fontsize=11, fontweight='bold',
            arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=8),
            color='#e74c3c')

# Resaltar el cuarto género sin atribuirle una tendencia temporal
genero_destacado = top_generos.index[3]
bars[3].set_color('#f39c12')  # Naranja para destacar
ax.annotate(f'{genero_destacado}\n4.º por frecuencia',
            xy=(top_generos.values[3], 3),
            xytext=(top_generos.values[3] + 150, 2.3),
            fontsize=10, fontweight='bold',
            arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=8),
            color='#f39c12')

# 3. KEY TAKEAWAY - Frase principal
plt.text(0.5, 0.92, f'Key Takeaway: {genero_top} aparece en {top_generos.iloc[0]:,} títulos del dataset.',
         transform=fig.transFigure, fontsize=11, style='italic',
         ha='center', color='#2c3e50',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

# 4. CALL TO ACTION - Qué hacer
plt.text(0.5, 0.87, 'Hipótesis a evaluar: contrastar este patrón con audiencia, costos y disponibilidad antes de decidir adquisiciones.',
         transform=fig.transFigure, fontsize=10, fontweight='bold',
         ha='center', color='#27ae60',
         bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.4))

# Configuración adicional
ax.set_yticks(range(len(top_generos)))
ax.set_yticklabels(top_generos.index, fontsize=11)
ax.set_xlabel('Cantidad de Títulos', fontsize=12, fontweight='bold')
ax.invert_yaxis()  # Invertir para que el #1 quede arriba

# Eliminar bordes innecesarios
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout(rect=[0, 0, 1, 0.85])  # Ajustar para texto superior
plt.show()
````

### 📖 Anatomía del Gráfico Narrativo

Analiza cada elemento del gráfico anterior:

**1. Headline (Título):**
> ❌ "Géneros en Netflix" (aburrido, no dice nada)
> ✅ Un título que muestre el género y porcentaje **calculados** por la celda (evidencia + número)

**2. Anotaciones (Flechas):**
> - La flecha roja resalta el género #1 y cuánto supera al #2
> - La flecha naranja destaca el cuarto género por frecuencia; no afirma que esté creciendo porque este gráfico no incluye tiempo

**3. Key Takeaway:**

## 📊 GRÁFICO 3: Gráfico de Líneas

⏱️ *Tiempo estimado: 10 minutos*

### ¿Qué es?
Un gráfico de líneas muestra **cómo cambia un valor a lo largo del tiempo**. Es perfecto para identificar tendencias y patrones temporales.

### ¿Cuándo usarlo?
- Analizar ventas mensuales o anuales
- Ver la evolución de métricas en el tiempo
- Identificar estacionalidad

> 💡 **Pregunta clave que responde:** "¿Cómo ha cambiado X en el tiempo?"

### Ejemplo: Crecimiento del Catálogo de Netflix

````python
# Preparar datos: Contenido agregado por año
# Convertir date_added a datetime
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'], errors='coerce')

# Extraer año de la fecha de adición
df_netflix['year_added'] = df_netflix['date_added'].dt.year

# Contar títulos agregados por año y tipo
agregados_por_año = df_netflix.groupby(['year_added', 'type']).size().unstack(fill_value=0)
agregados_por_año = agregados_por_año.dropna()

# Graficar
fig, ax = plt.subplots(figsize=(12, 6))

agregados_por_año.plot(kind='line', ax=ax, marker='o', linewidth=2, markersize=6)

plt.title('📈 Títulos registrados como agregados por año', fontsize=14, fontweight='bold')
plt.xlabel('Año')
plt.ylabel('Títulos Agregados')
plt.legend(title='Tipo de Contenido')
plt.grid(True, alpha=0.3)

# Agregar anotación en el pico
max_year = agregados_por_año['Movie'].idxmax()
max_value = agregados_por_año.loc[max_year, 'Movie']
plt.annotate(f'Pico: {max_value:,.0f} películas', 
             xy=(max_year, max_value), 
             xytext=(max_year-2, max_value+200),
             arrowprops=dict(arrowstyle='->', color='gray'),
             fontsize=10)

plt.tight_layout()
plt.show()
````

### 🎓 Interpretación

**Tendencias que debes comprobar en el gráfico:**
- Identifica el año y el valor del pico para cada línea; la anotación calcula el pico de películas.
- Compara años consecutivos solo donde haya datos: una caída en este dataset no demuestra una caída real del catálogo de Netflix.
- Describe cuál línea es mayor en los años observados, sin asumir que el patrón vale para todos los años.

**Insight de negocio (con límites):**
> "En **[año]** se registran **[número]** títulos de **[tipo]** agregados, frente a **[comparación]**. Las causas no se pueden establecer con este dataset; una hipótesis debe contrastarse con fuentes sobre licencias, producción y estrategia."

## 📊 GRÁFICO 4: Diagrama de Dispersión (Scatter Plot)

### ¿Qué es?
Un scatter plot muestra la **relación entre dos variables numéricas**. Cada punto representa una observación con sus valores en ambas variables.

### ¿Cuándo usarlo?
- Buscar correlaciones entre variables
- ¿Más precio = más ventas? ¿Más horas = mejor calificación?
- Identificar grupos o clusters

### Ejemplo: Año de Lanzamiento vs Duración de Películas

````python
# Preparar datos: extraer duración de películas
peliculas = df_netflix[df_netflix['type'] == 'Movie'].copy()
peliculas['duracion_min'] = peliculas['duration'].str.extract(r'(\d+)').astype(float)

# Filtrar valores válidos
peliculas = peliculas.dropna(subset=['duracion_min', 'release_year'])
peliculas = peliculas[peliculas['release_year'] >= 1980]

# Crear scatter plot
plt.figure(figsize=(12, 6))

plt.scatter(peliculas['release_year'], peliculas['duracion_min'], 
            alpha=0.3, color='steelblue', s=20)

# Agregar línea de tendencia
z = np.polyfit(peliculas['release_year'], peliculas['duracion_min'], 1)
p = np.poly1d(z)
plt.plot(peliculas['release_year'].sort_values(), 
         p(peliculas['release_year'].sort_values()), 
         "r--", linewidth=2, label='Tendencia')

plt.title('📊 Relación: Año de Lanzamiento vs Duración de Películas', fontsize=14, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Duración (minutos)')

# Agregar una referencia calculada: mediana de duración de esta muestra
# OJO: las líneas con `label` deben crearse ANTES de plt.legend(), si no no aparecen en ella
mediana_duracion = peliculas['duracion_min'].median()
plt.axhline(y=mediana_duracion, color='green', linestyle=':', alpha=0.7,
            label=f'Mediana: {mediana_duracion:.0f} min')

plt.legend()

plt.tight_layout()
plt.show()

# Calcular correlación
correlacion = peliculas['release_year'].corr(peliculas['duracion_min'])
print(f"\n📈 Correlación: {correlacion:.3f}")
print("   (Valores cercanos a 0 indican poca relación)")
````

### 🎓 Interpretación

**Observaciones que debes verificar:**
- Usa la correlación impresa: valores cercanos a 0 indican poca relación lineal, no ausencia total de relación.
- Describe el rango donde se concentra la nube de puntos según el gráfico, sin fijar un intervalo antes de verlo.
- Para afirmar más variabilidad en un periodo, compara formalmente rangos o desviaciones por grupos de años.

**Insight (basado en evidencia):**
> "La correlación calculada entre año y duración es **[valor]**. Esto resume una relación lineal en esta muestra, pero no permite concluir que el año cause una duración determinada."

## 📊 GRÁFICO 5: Boxplot (Diagrama de Caja)

### ¿Qué es?
Un boxplot muestra la **distribución de datos y sus valores atípicos (outliers)**. Muestra:
- La mediana (línea central)
- El rango intercuartílico (la caja)
- Los valores extremos (bigotes)
- Los outliers (puntos fuera de los bigotes)

### ¿Cuándo usarlo?
- Comparar distribuciones entre grupos
- Identificar valores atípicos
- Ver si los datos están sesgados

### Ejemplo: Duración de Películas por Clasificación

````python
# Preparar datos
top_ratings = ['TV-MA', 'TV-14', 'TV-PG', 'R', 'PG-13', 'PG']
peliculas_ratings = peliculas[peliculas['rating'].isin(top_ratings)].copy()

# Crear boxplot
plt.figure(figsize=(12, 6))

# Usar seaborn para un boxplot más atractivo
orden = peliculas_ratings.groupby('rating')['duracion_min'].median().sort_values(ascending=False).index

sns.boxplot(data=peliculas_ratings, x='rating', y='duracion_min',
            order=orden, hue='rating', hue_order=orden,
            palette='viridis', legend=False)

plt.title('📊 Distribución de Duración por Clasificación', fontsize=14, fontweight='bold')
plt.xlabel('Clasificación')
plt.ylabel('Duración (minutos)')

# Agregar una línea de referencia calculada en la muestra
mediana_general = peliculas_ratings['duracion_min'].median()
plt.axhline(y=mediana_general, color='red', linestyle='--', alpha=0.5,
            label=f'Mediana general: {mediana_general:.0f} min')
plt.legend()

plt.tight_layout()
plt.show()
````

### 🔍 Cómo Leer un Boxplot

```
                    ┌─────────────────┐
           Máximo ──│        │        │
                    │   ┌────┴────┐   │
              Q3 ───│   │         │   │── 75% de los datos
                    │   │─────────│   │
         Mediana ───│   │    ●    │   │── 50% (valor central)
                    │   │─────────│   │
              Q1 ───│   │         │   │── 25% de los datos
                    │   └────┬────┘   │
           Mínimo ──│        │        │
                    └─────────────────┘
              ●     ← Outliers (valores atípicos)
```

### 🎓 Interpretación

**Observaciones que debes verificar:**
- Compara las líneas de mediana de cada clasificación antes de decir cuál tiende a ser más larga o corta.
- Revisa las cajas y los bigotes: los puntos fuera de los bigotes son outliers según la regla del boxplot.
- Distingue una diferencia visual de una diferencia estadísticamente comprobada.

**Insight de negocio (con hipótesis explícita):**
